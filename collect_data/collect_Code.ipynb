### slm-research-vn/self-oss-instruct-sc2-exec-filter-50k-mistral-large

```python
ds_oss = load_dataset("slm-research-vn/self-oss-instruct-sc2-exec-filter-50k-mistral-large", split="train")

ds_oss
```

```python
def format_messages(example):
    messages = [
        {"role": "user", "content": example["instruction"]},
        {"role": "assistant", "content": example["responses"]}
    ]
    
    example["messages"] = messages
    
    return example

ds_oss = ds_oss.map(format_messages)
```

```python
ds_oss
```

```python
ds_oss_split = ds_oss.train_test_split(test_size=0.1, seed=42, shuffle=True)

ds_oss_split.push_to_hub("slm-research-vn/self-oss-instruct-sc2-exec_mistral-large-v1.0", token="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")
```
