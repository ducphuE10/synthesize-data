```python
from datasets import load_dataset, Dataset, concatenate_datasets
```

### microsoft/orca-math-word-problems-200k

```python
ds_orca = load_dataset("microsoft/orca-math-word-problems-200k", split="train")

ds_orca
```

```python
def format_messages(example):
    example["instruction"] = example["question"]
    messages = [
        {"role": "user", "content": example["question"]},
        {"role": "assistant", "content": example["answer"]}
    ]
    
    example["messages"] = messages
    
    return example

ds_orca = ds_orca.map(format_messages)
```

### hendrycks/competition_math

```python
ds_compe_math_dict = load_dataset("hendrycks/competition_math", trust_remote_code=True)

ds_compe_math_dict
```

```python
ds_compe_math = concatenate_datasets([ds_compe_math_dict["train"], ds_compe_math_dict["test"]])

def format_messages_compe_math(example):
    example["instruction"] = example["problem"]
    messages = [
        {"role": "user", "content": example["problem"]},
        {"role": "assistant", "content": example["solution"]}
    ]
    
    example["messages"] = messages
    
    return example

ds_compe_math = ds_compe_math.map(format_messages_compe_math)

ds_compe_math[0]
```

### TIGER-Lab/MathInstruct

```python
ds_math_instruct = load_dataset("TIGER-Lab/MathInstruct", split="train")

ds_math_instruct[0]
```

```python
def format_messages_math_instruct(example):
    messages = [
        {"role": "user", "content": example["instruction"]},
        {"role": "assistant", "content": example["output"]}
    ]
    
    example["messages"] = messages
    
    return example

ds_math_instruct = ds_math_instruct.map(format_messages_math_instruct)

ds_math_instruct[0]
```
